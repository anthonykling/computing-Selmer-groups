### Testing Corollary 6.1
If $b \in \mathbb{Z}[i]$ is fourth-power-free and in either of the following forms:

$$
b = i^{s_b} t^{t_b} p_1^{r_1} \cdots p_N^{r_N},
\qquad
s_b,t_b \in \{0,1,2,3\},
\qquad
r_i \in \{1,3\},
$$

where each $p_i \equiv 1 \pmod{t^7}$, or

$$
b = i^{s_b} t^{t_b} q_1^2 \cdots q_N^2,
\qquad
s_b,t_b \in \{0,2\},
$$

where each $q_j \equiv 1 \pmod{t^7}$, and

$$
R := \operatorname{rk}_{\mathbb{F}_2}(L(G_b)),
$$

then

$$
\#S^{(\varphi)}(E_b/\mathbb{Q}(i))
=
\begin{cases}
2^{N-R+2}
& \text{if } (s_b,t_b) \in \{(0,1),(1,2),(3,1)\}, \\[6pt]
2^{N-R+1}
& \text{otherwise}.
\end{cases}
$$

```python
def cor61(b):
    _, _, L = Laplacian(b)
    R = rank_f2(L)
    s_b = b.s()
    t_b = b.t()
    N = len(b.factor()) - 2
    if (s_b, t_b) in [(0,1),(1,2),(3,1)]:
        return 2**(N - R + 2)
    else:
        return 2**(N - R + 1)
```

```python
def random_b_corr61(norm_size, num_primes, s_b, s_t, parity, seed=None):

    if seed is not None:
        random.seed(seed)

    if parity not in ("even", "odd"):
        raise ValueError("parity must be either 'even' or 'odd'.")

    x = gaussian_primes(
        norm_size,
        GaussInt(1, 0),
        GaussInt(1, 1)**7
    )

    if len(x) < num_primes:
        raise ValueError("Not enough Gaussian primes found.")

    primes = random.sample(x, num_primes)

    b = GaussInt(1, 0)**s_b * GaussInt(1, 1)**s_t

    for p in primes:
        if parity == "even":
            exponent = 2
        else:
            exponent = random.choice([1, 3])

        b *= p**exponent

    return b
```
