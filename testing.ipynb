### Testing Corollary 6.1
If $b \in \mathbb{Z}[i]$ is fourth-power-free and in either of the following forms:

$$
b = i^{s_b} t^{t_b} p_1^{r_1} \cdots p_N^{r_N},
\qquad
s_b,t_b \in \{0,1,2,3\},
\qquad
r_i \in \{1,3\},
$$

where each $p_i \equiv 1 \pmod{t^7}$, or

$$
b = i^{s_b} t^{t_b} q_1^2 \cdots q_N^2,
\qquad
s_b,t_b \in \{0,2\},
$$

where each $q_j \equiv 1 \pmod{t^7}$, and

$$
R := \operatorname{rk}_{\mathbb{F}_2}(L(G_b)),
$$

then

$$
\#S^{(\varphi)}(E_b/\mathbb{Q}(i))
=
\begin{cases}
2^{N-R+2}
& \text{if } (s_b,t_b) \in \{(0,1),(1,2),(3,1)\}, \\[6pt]
2^{N-R+1}
& \text{otherwise}.
\end{cases}
$$

```python
def cor61(b):
    _, _, L = Laplacian(b)
    R = rank_f2(L)
    s_b = b.s()
    t_b = b.t()
    N = len(b.factor()) - 2
    if (s_b, t_b) in [(0,1),(1,2),(3,1)]:
        return 2**(N - R + 2)
    else:
        return 2**(N - R + 1)
```

```python
def random_b_corr61(norm_size, num_primes, s_b, s_t, parity, seed=None):
    '''This generates a random Gaussian prime with num_primes number of odd primary prime factors which satisfy Corollary 6.1.
    Each prime has norm less than norm_size with specified values of s_b and s_t.
    Parity refers to which case of Corollary 6.1 is satisfied, either "even" or "odd".  
    '''
    if seed is not None:
        random.seed(seed)

    if parity not in ("even", "odd"):
        raise ValueError("parity must be either 'even' or 'odd'.")

    x = gaussian_primes(
        norm_size,
        GaussInt(1, 0),
        GaussInt(1, 1)**7
    )

    if len(x) < num_primes:
        raise ValueError("Not enough Gaussian primes found.")

    primes = random.sample(x, num_primes)

    b = GaussInt(1, 0)**s_b * GaussInt(1, 1)**s_t

    for p in primes:
        if parity == "even":
            exponent = 2
        else:
            exponent = random.choice([1, 3])

        b *= p**exponent

    return b
```

```python
# Testing the odd case.
num_primes = 4
num_cases = 10
for s_b in [0, 1, 2, 3]:
    for s_t in [0, 1, 2, 3]:
            for i in range(num_primes):
                for j in range(num_cases):
                    b = random_b_corr61(10000, num_primes, s_b, s_t, "odd")
                    assert len(Selmer_group(b)) == cor61(b), f"Failed for b={b}, s_b={s_b}, s_t={s_t}"
                    if j % 10 == 0:
                        print(f"Test passed for b={b}, s_b={s_b}, s_t={s_t}")
```

### Theorem 6.8

Let $b \in \mathbb{Z}[i]$ be fourth-power free so that it can be uniquely factored
$$ b = i^{s_b}  (1+i)^{t_b}  \mathfrak{p}_1^{r_1} \cdots \mathfrak{p}_M^{r_M} \mathfrak{q}_1^{2}\cdots\mathfrak{q}_N^{2}$$
for some $s_b,t_b\in \{0,1,2,3\}, r_i\in\{1,3\}$ with the $\mathfrak{p}_i,\mathfrak{q}_j$ distinct primary Gaussian primes.

Suppose each $\mathfrak{p}_i$ and $\mathfrak{q}_j$ is inert, lying over rational primes $p_i$ and $q_j$, respectively. For $k\in\{3,7,11,15\}$, set

$$
M_k:=\#\{p_i:p_i\equiv k\pmod{16}\},
$$

$$
N_k:=\#\{q_j:q_j\equiv k\pmod{16}\},
$$

so that $M=M_3+M_7+M_{11}+M_{15}$, and similarly for $N$. Then the following hold.

(1) If $t_b$ is odd, then

$$
S^{(\varphi)}(E_b/\mathbb{Q}(i))
\cong
(\mathbb{Z}/2\mathbb{Z})^{M_7+M_{15}+N_{15}+c},
$$

where

$$
c=
\begin{cases}
2 & \text{if } (s_b,t_b)\in\{(0,1),(3,1)\},\\
1 & \text{otherwise}.
\end{cases}
$$

(2) If $(s_b,t_b)\in\{(1,0),(3,0),(0,2),(2,2)\}$, then

$$
S^{(\varphi)}(E_b/\mathbb{Q}(i))
\cong
(\mathbb{Z}/2\mathbb{Z})^{M+N_7+N_{15}+c},
$$

where

$$
c=
\begin{cases}
1 & \text{if } t_b=0,\\
-1 & \text{if } t_b=2 \text{ and } M_3+M_{11}
    \text{ is positive and even},\\
1 & \text{if } t_b=2,\ M_3=M_{11}=0,\text{ and } M_7
    \text{ is even},\\
0 & \text{otherwise}.
\end{cases}
$$

(3) If $(s_b,t_b)\in\{(0,0),(2,0),(1,2),(3,2)\}$, then

$$
S^{(\varphi)}(E_b/\mathbb{Q}(i))
\cong
(\mathbb{Z}/2\mathbb{Z})^{M+N+c},
$$

where

$$
c=
\begin{cases}
2 & \text{if } s_b=1 \text{ and } M_3=M_{11}=N_3=N_{11}=0,\\
1 & \text{if } s_b=3,\text{ or } s_b=0 \text{ and } M_3+M_{11}
    \text{ is even},\\
0 & \text{if } s_b=t_b \text{ and } m_b\equiv 0,3\pmod{4},\\
0 & \text{otherwise}.
\end{cases}
$$

```python
def MN(k, b):
    '''This function computes M_k and N_k as in Theorem 6.8.
    Note that b must be a product of inert primes.  If p is an inert primary prime, then it lies over the rational prime -p.
    Note that -p is congruent to 3 modulo 4.
    '''
    M_odd_factors = [p for p, e in b.factor().items() if p != GaussInt(0,1) and p != GaussInt(1,1) and e % 2 == 1]
    N_odd_factors = [p for p, e in b.factor().items() if p != GaussInt(0,1) and p != GaussInt(1,1) and e == 2]
    if k == 0:
        return len(M_odd_factors), len(N_odd_factors)
    
    M_count = 0
    N_count = 0
    for p in M_odd_factors:
        if GaussInt(16,0).divides(-p - GaussInt(k,0)):
            M_count += 1
    for q in N_odd_factors:
        if GaussInt(16,0).divides(-q - GaussInt(k,0)):
            N_count += 1
    return M_count, N_count
        
```
